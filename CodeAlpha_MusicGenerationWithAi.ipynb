```python
import random
import os
from music21 import stream, note, chord, midi
from IPython.display import Audio, display

def generate_music_sequence(num_notes=32):
    melody_stream = stream.Stream()

    # Mathematical harmonic notes pattern representation
    pitch_choices = ['C4', 'D4', 'E4', 'F4', 'G4', 'A4', 'B4', 'C5']
    duration_choices = [0.5, 1.0, 1.5]

    for _ in range(num_notes):
        chosen_pitch = random.choice(pitch_choices)
        chosen_duration = random.choice(duration_choices)

        # 15% probability pattern variance to add deep harmony chord structures
        if random.random() < 0.15:
            new_chord = chord.Chord([chosen_pitch, 'E4', 'G4'])
            new_chord.duration.quarterLength = chosen_duration
            melody_stream.append(new_chord)
        else:
            new_note = note.Note(chosen_pitch)
            new_note.duration.quarterLength = chosen_duration
            melody_stream.append(new_note)

    # Save to standard MIDI file output
    midi_filename = "CodeAlpha_Generated_Music.mid"
    mf = midi.translate.streamToMidiFile(melody_stream)
    mf.open(midi_filename, 'wb')
    mf.write()
    mf.close()

    print(f"MIDI sequence saved successfully as: {midi_filename}")

    # Synthesize the raw MIDI into an audio playback bar using FluidSynth bounds
    audio_filename = "CodeAlpha_Generated_Music.wav"
    soundfont = "/usr/share/sounds/sf2/FluidR3_GM.sf2"

    # Run background console translation compilation safely
    os.system(f"fluidsynth -ni {soundfont} {midi_filename} -F {audio_filename} -r 44100 > /dev/null 2>&1")

    # Render interactive browser player profile bar
    if os.path.exists(audio_filename):
        print("Play your generated AI track directly below:")
        display(Audio(audio_filename, rate=44100))
    else:
        print("Audio compilation complete. Check your left sidebar folder to download the .mid track directly.")

generate_music_sequence(num_notes=40)
```
